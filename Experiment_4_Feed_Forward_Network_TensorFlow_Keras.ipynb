## 1. Theory

A **Feed-Forward Neural Network (FFNN)** is a neural network in which information moves in one direction: from the input layer to the hidden layer(s) and finally to the output layer.

Unlike recurrent networks, an FFNN does not send information back to an earlier layer.

### Basic Structure

```text
Input Layer → Hidden Layer → Output Layer
   ↓              ↓               ↓
Features       Neurons          Class
```

In this experiment, the **Iris dataset** is used. The model learns to classify flowers into three species using four measurements:

- Sepal length
- Sepal width
- Petal length
- Petal width

The three classes are represented as numbers 0, 1 and 2.

````python
# Step 1: Import libraries
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

print("TensorFlow version:", tf.__version__)
````

````python
# Step 2: Load the Iris dataset
iris = load_iris()
X = iris.data
y = iris.target

print("Input shape:", X.shape)
print("Target shape:", y.shape)
print("Class names:", iris.target_names)
````

````python
# Step 3: Split the dataset into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Standardize the input features
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
````

## 2. Build the Feed-Forward Neural Network

The network used here contains:

- **Input layer:** 4 features
- **Hidden layer 1:** 16 neurons with ReLU activation
- **Hidden layer 2:** 8 neurons with ReLU activation
- **Output layer:** 3 neurons with Softmax activation

Softmax gives a probability for each of the three flower classes.

````python
# Step 4: Create the Feed-Forward Neural Network
model = keras.Sequential([
    layers.Input(shape=(4,)),
    layers.Dense(16, activation="relu"),
    layers.Dense(8, activation="relu"),
    layers.Dense(3, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
````

### Explanation of the Model

- `Dense(16)` creates the first hidden layer with 16 neurons.
- `Dense(8)` creates the second hidden layer with 8 neurons.
- `Dense(3)` creates three output neurons because there are three flower classes.
- `relu` is used in the hidden layers.
- `softmax` is used in the output layer to obtain class probabilities.
- `adam` is used as the optimizer.
- `sparse_categorical_crossentropy` is used because the target classes are integer labels.

````python
# Step 5: Train the Feed-Forward Neural Network
history = model.fit(
    X_train,
    y_train,
    validation_split=0.2,
    epochs=50,
    batch_size=16,
    verbose=0
)

print("Model training completed.")
````

````python
# Step 6: Evaluate the model
loss, accuracy = model.evaluate(X_test, y_test, verbose=0)

print("Test Loss:", round(float(loss), 4))
print("Test Accuracy:", round(float(accuracy * 100), 2), "%")
````

````python
# Step 7: Predict the test data
probabilities = model.predict(X_test, verbose=0)
predicted_classes = np.argmax(probabilities, axis=1)

print("Actual classes:   ", y_test[:15])
print("Predicted classes:", predicted_classes[:15])
````

````python
# Step 8: Display a few predictions with flower names
for i in range(10):
    actual = iris.target_names[y_test[i]]
    predicted = iris.target_names[predicted_classes[i]]
    confidence = np.max(probabilities[i]) * 100
    print(f"Actual: {actual:10s} | Predicted: {predicted:10s} | Confidence: {confidence:.2f}%")
````

````python
# Step 9: Plot training and validation accuracy
import matplotlib.pyplot as plt

plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("FFNN Training and Validation Accuracy")
plt.legend()
plt.show()
````

````python
# Step 10: Plot training and validation loss
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("FFNN Training and Validation Loss")
plt.legend()
plt.show()
````
