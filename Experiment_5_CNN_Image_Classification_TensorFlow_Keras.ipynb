## 1. Theory

A **Convolutional Neural Network (CNN)** is a deep learning model specially designed for processing images.

A typical CNN contains:

1. **Convolution layer** – detects useful features such as edges and patterns.
2. **Pooling layer** – reduces the size of feature maps while retaining important information.
3. **Flatten layer** – converts feature maps into a one-dimensional vector.
4. **Dense layer** – performs classification using the extracted features.
5. **Output layer** – gives the predicted class.

### CNN Flow

```text
Image
  ↓
Convolution
  ↓
ReLU
  ↓
Max Pooling
  ↓
Convolution
  ↓
ReLU
  ↓
Max Pooling
  ↓
Flatten
  ↓
Dense
  ↓
Output
```

In this experiment, the **CIFAR-10 dataset** is used. It contains 10 image classes such as airplane, automobile, bird, cat, deer, dog, frog, horse, ship and truck.

````python
# Step 1: Import required libraries
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

print("TensorFlow version:", tf.__version__)
````

````python
# Step 2: Load the CIFAR-10 dataset
(x_train, y_train), (x_test, y_test) = keras.datasets.cifar10.load_data()

class_names = [
    "airplane", "automobile", "bird", "cat", "deer",
    "dog", "frog", "horse", "ship", "truck"
]

print("Training images:", x_train.shape)
print("Testing images:", x_test.shape)
print("Number of classes:", len(class_names))
````

````python
# Step 3: Display sample images
plt.figure(figsize=(10, 6))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i])
    plt.title(class_names[int(y_train[i])])
    plt.axis("off")
plt.tight_layout()
plt.show()
````

## 2. Preprocess the Images

Pixel values in an image normally range from **0 to 255**. Neural networks generally work better when the values are scaled to a smaller range.

We divide the pixel values by 255 so that they lie between **0 and 1**.

````python
# Step 4: Normalize image pixel values
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

print("Minimum pixel value:", x_train.min())
print("Maximum pixel value:", x_train.max())
````

## 3. Build the CNN Model

The CNN used in this experiment contains:

- Two convolution layers
- Two max-pooling layers
- One flatten layer
- One dense hidden layer
- One output layer with 10 neurons

Because CIFAR-10 has 10 classes, the output layer has 10 neurons.

````python
# Step 5: Create the CNN model
model = keras.Sequential([
    layers.Input(shape=(32, 32, 3)),
    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation="relu"),
    layers.Dense(10, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
````

### Explanation of Important Layers

- `Conv2D` performs convolution to extract image features.
- `ReLU` introduces non-linearity and helps the network learn complex patterns.
- `MaxPooling2D` reduces the spatial size of the feature maps.
- `Flatten` converts the 2D feature maps into a 1D vector.
- `Dense` learns relationships between extracted features.
- `Softmax` produces probabilities for the 10 classes.

````python
# Step 6: Train the CNN
history = model.fit(
    x_train,
    y_train,
    validation_split=0.1,
    epochs=10,
    batch_size=64,
    verbose=1
)
````

````python
# Step 7: Evaluate the CNN on test images
test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)

print("Test Loss:", round(float(test_loss), 4))
print("Test Accuracy:", round(float(test_accuracy * 100), 2), "%")
````

````python
# Step 8: Predict test images
probabilities = model.predict(x_test[:10], verbose=0)
predicted_classes = np.argmax(probabilities, axis=1)

for i in range(10):
    actual = class_names[int(y_test[i][0])]
    predicted = class_names[int(predicted_classes[i])]
    confidence = float(np.max(probabilities[i]) * 100)
    print(f"Image {i+1}: Actual = {actual:10s} | Predicted = {predicted:10s} | Confidence = {confidence:.2f}%")
````

````python
# Step 9: Display sample predictions
plt.figure(figsize=(10, 6))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_test[i])
    actual = class_names[int(y_test[i][0])]
    predicted = class_names[int(predicted_classes[i])]
    plt.title(f"A: {actual}\nP: {predicted}", fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
````

````python
# Step 10: Plot training and validation accuracy
plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("CNN Training and Validation Accuracy")
plt.legend()
plt.show()
````

````python
# Step 11: Plot training and validation loss
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("CNN Training and Validation Loss")
plt.legend()
plt.show()
````
