## 1. Theory

A vector is a one-dimensional collection of numbers. In TensorFlow, vectors are represented using tensors.

If `A = [1, 2, 3, 4]` and `B = [5, 6, 7, 8]`, vector addition is performed element by element:

`A + B = [1+5, 2+6, 3+7, 4+8] = [6, 8, 10, 12]`.


````python
# Step 1: Import TensorFlow
import tensorflow as tf

print("TensorFlow version:", tf.__version__)
````

````python
# Step 2: Create two vectors
A = tf.constant([1, 2, 3, 4])
B = tf.constant([5, 6, 7, 8])

print("Vector A:", A.numpy())
print("Vector B:", B.numpy())
````

````python
# Step 3: Perform vector addition
C = tf.add(A, B)
print("A + B =", C.numpy())
````

````python
# Step 4: Verify using the + operator
C2 = A + B
print("Using + operator:", C2.numpy())
print("Both methods give the same result:", tf.reduce_all(tf.equal(C, C2)).numpy())
````

## Expected Output

```text
Vector A: [1 2 3 4]
Vector B: [5 6 7 8]
A + B = [ 6  8 10 12]
Using + operator: [ 6  8 10 12]
Both methods give the same result: True
```
