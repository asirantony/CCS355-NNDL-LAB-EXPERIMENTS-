## 1. Theory

A **pre-trained model** is a neural network that has already been trained on a large dataset. Instead of training a model from the beginning, we can use its learned knowledge to classify new images.

In this experiment, we use **MobileNetV2**, a lightweight convolutional neural network available in Keras. It is pre-trained on the **ImageNet** dataset, which contains many categories of real-world objects.

### Basic Flow

```text
Given Image
     ↓
Resize Image
     ↓
Preprocessing
     ↓
Pre-trained MobileNetV2
     ↓
ImageNet Predictions
     ↓
Top-5 Classes + Confidence
```

### Important Point

The model predicts classes from the categories it learned during ImageNet training. Therefore, the model may not correctly classify an object that is outside those categories.

````python
# Step 1: Import required libraries
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input, decode_predictions
import numpy as np
import matplotlib.pyplot as plt

print("TensorFlow version:", tf.__version__)
````

## 2. Load the Pre-trained Model

`weights="imagenet"` loads weights learned from the ImageNet dataset.

`include_top=True` keeps the final classification layer, allowing the model to directly predict ImageNet classes.

````python
# Step 2: Load MobileNetV2 with ImageNet weights
model = MobileNetV2(
    weights="imagenet",
    include_top=True
)

print("Pre-trained MobileNetV2 loaded successfully.")
````

## 3. Upload a Given Image

Run the following cell and select an image from your computer.

**Suggested images:** dog, cat, bird, car, bicycle, fruit, bottle, or another common ImageNet object.

````python
# Step 3: Upload an image in Google Colab
from google.colab import files

uploaded = files.upload()

image_path = next(iter(uploaded.keys()))
print("Selected image:", image_path)
````

````python
# Step 4: Load and display the image
image = keras.utils.load_img(image_path, target_size=(224, 224))

plt.figure(figsize=(6, 6))
plt.imshow(image)
plt.axis("off")
plt.title("Input Image")
plt.show()
````

## 4. Preprocess the Image

MobileNetV2 expects an input image of size **224 × 224 pixels**.

The `preprocess_input()` function converts the image values into the format expected by MobileNetV2.

````python
# Step 5: Convert image to an array and preprocess it
image_array = keras.utils.img_to_array(image)
image_array = np.expand_dims(image_array, axis=0)
image_array = preprocess_input(image_array)

print("Preprocessed image shape:", image_array.shape)
````

## 5. Predict the Image Class

The model produces probabilities for many ImageNet classes. We decode the predictions and display the top 5 classes.

````python
# Step 6: Make predictions
predictions = model.predict(image_array, verbose=0)

# Decode the top 5 predictions
results = decode_predictions(predictions, top=5)[0]

print("Top 5 Predictions:\n")
for rank, (_, class_name, probability) in enumerate(results, start=1):
    print(f"{rank}. {class_name:25s} : {probability * 100:.2f}%")
````

````python
# Step 7: Display the best prediction
best_class = results[0][1]
best_probability = results[0][2]

print("Predicted Class:", best_class)
print("Confidence:", f"{best_probability * 100:.2f}%")
````

````python
# Step 8: Visualize the top 5 predictions
labels = [item[1] for item in results]
probabilities = [item[2] * 100 for item in results]

plt.figure(figsize=(9, 5))
plt.bar(labels, probabilities)
plt.ylabel("Confidence (%)")
plt.xlabel("Predicted Class")
plt.title("Top-5 ImageNet Predictions")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
````

## 6. Understanding the Prediction

Suppose the model produces:

```text
1. golden_retriever   : 87.50%
2. Labrador_retriever :  8.20%
3. kuvasz             :  1.50%
...
```

The model's top prediction is **golden retriever** with a confidence score of 87.50%.

**Confidence is not the same as guaranteed correctness.** It represents the model's predicted probability for that class.
